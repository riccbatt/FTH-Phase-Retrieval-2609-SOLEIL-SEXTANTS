# Multimode phase retrieval
Minimal use of `phase_retrieval_core_multimode.py`.

```python
import numpy as np
from library import phase_retrieval_core_multimode as pr
```

```python
pos_hologram = np.load("data/pos_hologram.npy")
neg_hologram = np.load("data/neg_hologram.npy")
mask_pixel = np.load("data/mask_pixel.npy")
supportmask = np.load("data/supportmask.npy")
```

```python
recipe = {
    # Parallel lists: every index defines one reconstruction stage.
    "algorithm_list": ["HAPRE", "ER"],  # Support-projection algorithm for each stage.
    "number_iterations": [700, 50],       # Iterations performed by each stage.
    "helicity": ["pos", "pos"],         # Input hologram updated by each stage.
    "beta_zero": [0.5, 0.5],             # Beta-schedule starting/reference values.
    "beta_mode": ["arctan", "const"],   # Beta schedule names or arrays.
    "alpha_zero": [0.0, 0.0],            # TV-step strength; zero disables TV.
    "alpha_mode": ["const", "const"],   # Alpha schedule names or arrays.
    "RL_its": [0, 0],                    # Richardson-Lucy iterations; zero means full coherence.
    "RL_freqs": [1e9, 1e9],              # RL update interval; values above Nit disable RL.
    "TV_freqs": [1e9, 1e9],              # TV update interval.
    "plot_every": [350, 25],              # Error sampling/plot interval.
    "average_img": [30, 30],              # Number of best late iterates to average.
    "Fourier_last": [True, True],         # Reapply the Fourier constraint at each stage end.
    "Nmodes": 3,                         # Number of mutually incoherent reconstructed modes.
    "hologram_intensity_cutoff_vmin": -1, # Percentile baseline subtraction; negative disables it.
    "Startimage": [None, "pos"],         # None, a helicity label, or explicit modal fields.
    "Startgamma": [None, None],           # Initial partial-coherence kernels.
}

results = pr.phase_retrieval_algorithm(
    pos_hologram,
    neg_hologram,
    mask_pixel,
    supportmask,
    phase_retrieval_recipe=recipe,
)
```
