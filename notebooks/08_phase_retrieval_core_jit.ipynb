Minimal direct use of `phase_retrieval_core_jit.py`.

```python
import numpy as np
from library import phase_retrieval_core_jit as pr
```

```python
measured_amplitude = np.load("data/measured_amplitude.npy")
supportmask = np.load("data/supportmask.npy")
mask_pixel = np.load("data/mask_pixel.npy")
start_field = np.load("data/start_field.npy")

print(pr.cuda_jit_status())
# pr.warm_up_jit(measured_amplitude.shape, modes=["HAPRE"])
```

```python
# PhaseRtrv_core_jit uses a function-argument recipe rather than a separate default recipe.
recipe = {
    "mode": "HAPRE",       # ER, SF, HAPRE, RAAR, HIOs, HIO, CHIO, or HPR; OSS falls back.
    "Nit": 700,             # Number of iterations.
    "beta_zero": 0.5,       # Beta-schedule starting/reference value.
    "beta_mode": "arctan", # Beta schedule name or a NumPy array of length Nit.
    "alpha_zero": 0.0,      # TV strength; nonzero active TV currently falls back.
    "alpha_mode": "const", # Alpha schedule name or array.
    "seed": False,          # Use deterministic NumPy initialization when Phase is None.
    "plot_every": 20,       # Error sampling interval.
    "real_object": False,   # Reserved compatibility option; complex objects are standard.
    "average_img": 30,      # Number of best late iterates to average.
    "Fourier_last": True,   # Reapply measured amplitudes before returning.
    "gamma": None,          # Initial coherence kernel; active RL currently falls back.
    "RL_freq": None,        # Richardson-Lucy update interval.
    "RL_it": 0,             # Richardson-Lucy iterations per update; zero disables RL.
    "TV_freq": 1e9,         # TV update interval.
    "fallback": False,      # False requires CUDA JIT; True permits ordinary fallback.
}

field, diffraction_error, support_error, gamma = pr.PhaseRtrv_core_jit(
    diffract=measured_amplitude,
    mask=supportmask,
    bsmask=mask_pixel,
    Phase=start_field,
    **recipe,
)
```
