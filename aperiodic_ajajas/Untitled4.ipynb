```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
l=2.7e-9
w=300e-9
theta=np.arcsin(l/w)

D=125e-3
pxsize=11e-6

np.tan(theta)*D/pxsize
```

```python
a=np.zeros((1000,1000))

for i in range(20):
    a[i::40]=1

fig,ax=plt.subplots()
ax.imshow(a)

b=np.abs(np.fft.fftshift(np.fft.fft2(a)))**2


fig,ax=plt.subplots()
ax.imshow(b)

fig,ax=plt.subplots()
ax.plot(b[500:,500])
ax.set_yscale("log")
ax.set_ylim(1e8,1e20)

ax.set_xlim(0,200)
```
